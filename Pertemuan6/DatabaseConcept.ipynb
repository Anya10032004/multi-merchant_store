### Database Relationships (Relasi Database)

A database relationship is a relationship between two tables that explains how the data in one table is connected to the data in another table.

For example, a database might have a Mahasiswa table and a MataKuliah table. Relationships help us connect the data between them.

There are 3 main types of relationships:
- One to One(1:1)
- One-to-Many (1:N)
- Many-to-Many (M:N)

#### One-to-One (1:1)

Definition: 

One record in Table A is related to at most one record in Table B, and vice versa.

Example:
- One person has one SIM (in this simplified example).
- One SIM belongs to one person.
How to understand it: Both sides have a maximum of one relationship.

#### One-to-Many (1:N)

Definition: 

One record in Table A can be related to many records in Table B, but each record in Table B is related to at most one record in Table A.

Example:

- One lecturer supervises many students.

- Each student has one primary supervisor in this simplified example.

How to understand it: One lecturer can connect to multiple students, but each student has only one primary supervisor

#### Many-to-Many (M:N)

Definition: 

One record in Table A can be related to many records in Table B, and one record in Table B can also be related to many records in Table A.

Example:

- One student can take Database, Python, and Statistics.

- One course, such as Database, can be taken by many students.

How to understand it: 
- Both sides can have multiple relationships. 
How do we implement it in a database?
- We usually create a third table, called a junction table (or associative table), to connect the two tables.
